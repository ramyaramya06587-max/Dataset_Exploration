```python
import pandas as pd

# Step 1: Load the Iris dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

print("TASK 17: DATASET EXPLORATION")
print("=" * 50)

# Step 2: Display the first five records
print("\n1. First Five Records:")
print(df.head())

# Step 3: Display the last five records
print("\n2. Last Five Records:")
print(df.tail())

# Step 4: Check dataset dimensions
print("\n3. Dataset Dimensions:")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)

# Step 5: Display column names
print("\n4. Column Names:")
print(df.columns.tolist())

# Step 6: Display dataset information
print("\n5. Dataset Information:")
df.info()

# Step 7: Identify numerical columns
print("\n6. Numerical Columns:")
numerical_columns = df.select_dtypes(include="number").columns.tolist()
print(numerical_columns)

# Step 8: Identify categorical columns
print("\n7. Categorical Columns:")
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()
print(categorical_columns)

# Step 9: Generate descriptive statistics
print("\n8. Descriptive Statistics:")
print(df.describe())

# Step 10: Check missing values
print("\n9. Missing Values:")
print(df.isnull().sum())

# Step 11: Count unique values in each column
print("\n10. Unique Values in Each Column:")
print(df.nunique())

# Step 12: Count flowers in each species
print("\n11. Species Distribution:")
print(df["species"].value_counts())

# Step 13: Calculate average measurements by species
print("\n12. Average Measurements by Species:")
print(df.groupby("species")[numerical_columns].mean().round(2))

# Step 14: Display at least five observations
print("\n13. OBSERVATIONS")
print("=" * 50)

print("1. The dataset contains", df.shape[0], "rows and",
      df.shape[1], "columns.")

print("2. The dataset contains", len(numerical_columns),
      "numerical columns and", len(categorical_columns),
      "categorical columns.")

print("3. The dataset contains", df["species"].nunique(),
      "different Iris flower species.")

print("4. Total missing values:", df.isnull().sum().sum())

print("5. Average measurements differ across flower species.")
print(df.groupby("species")[numerical_columns].mean().round(2))

print("\nDataset exploration completed successfully!")
```
